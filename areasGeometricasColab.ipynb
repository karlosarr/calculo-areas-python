# Calculo de areas geometricas

Este notebook implementa y prueba seis funciones de calculo de areas.
Las pruebas muestran como validar casos normales, limites y errores de entrada.

```python
from pathlib import Path
import importlib

try:
    import areasGeometricas
except ModuleNotFoundError:
    from google.colab import files

    print("Sube el archivo areasGeometricas.py para continuar.")
    files.upload()
    importlib.invalidate_caches()

    if not Path("areasGeometricas.py").is_file():
        raise FileNotFoundError("No se encontro areasGeometricas.py.")

    import areasGeometricas

from areasGeometricas import (
    areaCirculo,
    areaCuadrado,
    areaRectangulo,
    areaRombo,
    areaTrapecio,
    areaTriangulo,
)
```

## Ejemplos de uso

```python
# Cambia estos valores para probar otras dimensiones.

# Triangulo: base y altura.
baseTriangulo = 10
alturaTriangulo = 4

# Cuadrado: longitud de un lado.
ladoCuadrado = 5

# Rectangulo: base y altura.
baseRectangulo = 8
alturaRectangulo = 3

# Circulo: longitud del radio.
radioCirculo = 3

# Trapecio: base mayor, base menor y altura.
baseMayorTrapecio = 10
baseMenorTrapecio = 6
alturaTrapecio = 4

# Rombo: diagonal mayor y diagonal menor.
diagonalMayorRombo = 10
diagonalMenorRombo = 6

print("Todas las dimensiones estan expresadas en centimetros (cm).")
print("Triangulo: {:.2f} cm^2".format(areaTriangulo(baseTriangulo, alturaTriangulo)))
print("Cuadrado: {:.2f} cm^2".format(areaCuadrado(ladoCuadrado)))
print("Rectangulo: {:.2f} cm^2".format(areaRectangulo(baseRectangulo, alturaRectangulo)))
print("Circulo: {:.2f} cm^2".format(areaCirculo(radioCirculo)))
print("Trapecio: {:.2f} cm^2".format(areaTrapecio(baseMayorTrapecio, baseMenorTrapecio, alturaTrapecio)))
print("Rombo: {:.2f} cm^2".format(areaRombo(diagonalMayorRombo, diagonalMenorRombo)))
```
